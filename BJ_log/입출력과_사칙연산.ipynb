```python
'''문제 1000번
두 정수 A와 B를 입력받은 다음, A+B를 출력하는 프로그램을 작성하시오.

입력
첫째 줄에 A와 B가 주어진다. (0 < A, B < 10)

출력
첫째 줄에 A+B를 출력한다.'''
A,B = map(int, input().split())
print(A+B)
```

여기서는 두 숫자를 한번에 입력받느라 map과 split을 사용했다.
input()으로 입력받은 문자열을 split()으로 나누고, map()으로 각각을 int로 변환했다.
map()은 첫 번째 인자로 함수를 받고, 두 번째 인자로 반복 가능한 객체를 받는다.
map()은 반복 가능한 객체의 각 요소에 함수를 적용하여 새로운 반복 가능한 객체를 반환한다. 여기서는 int 함수를 A와 B에 적용하여 각각을 정수로 변환했다.
map()과 split()은 숫자 두개 이상도 사용 가능하다! 리스트로 반환하기 때문에 여러 숫자를 입력받을 때 유용하다.
split()은 문자열을 공백을 기준으로 나누어 리스트로 반환한다. 여기서는 input()으로 입력받은 문자열을 공백으로 나누어 A와 B를 얻었다.

```python
'''문제 1001번
두 정수 A와 B를 입력받은 다음, A-B를 출력하는 프로그램을 작성하시오.

입력
첫째 줄에 A와 B가 주어진다. (0 < A, B < 10)

출력
첫째 줄에 A-B를 출력한다.
'''
A,B = map(int, input().split())
print(A-B)
```

```python
'''문제 10998번
두 정수 A와 B를 입력받은 다음, A×B를 출력하는 프로그램을 작성하시오.

입력
첫째 줄에 A와 B가 주어진다. (0 < A, B < 10)

출력
첫째 줄에 A×B를 출력한다.
'''
A,B = map(int, input().split())
print(A*B)
```

```python
'''문제 1008번
두 정수 A와 B를 입력받은 다음, A/B를 출력하는 프로그램을 작성하시오.

입력
첫째 줄에 A와 B가 주어진다. (0 < A, B < 10)

출력
첫째 줄에 A/B를 출력한다. 실제 정답과 출력값의 절대오차 또는 상대오차가 10-9 이하이면 정답이다.

'''
A,B = map(int, input().split())
print(A/B)
```

```python
'''문제 10869번
두 자연수 A와 B가 주어진다. 이때, A+B, A-B, A*B, A/B(몫), A%B(나머지)를 출력하는 프로그램을 작성하시오. 

입력
두 자연수 A와 B가 주어진다. (1 ≤ A, B ≤ 10,000)

출력
첫째 줄에 A+B, 둘째 줄에 A-B, 셋째 줄에 A*B, 넷째 줄에 A/B, 다섯째 줄에 A%B를 출력한다.'''
A,B = map(int, input().split())
print(A+B)
print(A-B)
print(A*B)
print(A//B)
print(A%B)
```
