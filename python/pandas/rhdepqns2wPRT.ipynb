1. 아래와 같은 series sr1을 만드시오
 Index: a,b,c,d,e
 Data: 0~14까지의 random value 5개
2. sr1에서 두번째로 큰 data의 값과 index를 구하시오

```python
# numpy의 randint 함수: 지정된 범위에서 랜덤 정수 생성
# np.random.randint(0,15,size=5) → 0부터 14까지의 정수 중 5개를 랜덤으로 선택
# index=['a','b','c','d','e'] → 각 데이터값에 해당하는 인덱스(라벨) 지정
sr1 = pd.Series(np.random.randint(0,15,size=5), index=['a','b','c','d','e'])
sr1

# pandas의 sort_values 함수: Series나 DataFrame의 값을 기준으로 정렬
# ascending=False → 내림차순 정렬 (큰 값부터 작은 값 순서), True면 오름차순
# sr1.sort_values(ascending=False) → sr1의 값들을 큰 순서대로 재배열하여 새 Series 반환
sr2 = sr1.sort_values(ascending=False)
sr2

# Series의 index 속성: Series의 인덱스(라벨)들을 Index 객체로 반환
# 정렬된 sr2의 인덱스 순서 확인 (값이 큰 순서대로 인덱스도 재배열됨)
sr2.index

# 인덱스 접근: index[1] → Index 객체의 두 번째 요소 (0부터 시작)
# 정렬된 Series에서 두 번째로 큰 값의 인덱스(라벨)를 가져옴
sr2.index[1]

# iloc 속성: integer location의 약자로, 정수 인덱스를 이용해 데이터 접근
# iloc[1] → Series의 두 번째 위치(0부터 시작)에 있는 실제 데이터 값
# 정렬된 Series에서 두 번째로 큰 실제 데이터 값을 가져옴
sr2.iloc[1]
```

pandas_data3.txt를 읽어서 다음을 수행하시오
1. final이 40점 이상인 학생의 수
2. 1학년 학생의 기말 점수. 그리고 그 점수들의 평균
3. proj가 100점인 학생의 mid, final, att 점수
4. att이 5점이상 이고 proj가 90점 이상인 학생 name
5. “mid_grade” column 만들어, mid 점수가 25점 이상 A, 20점 이상 B, 나머지
는 C를 넣으시오
6. gender 별 기말점수의 평균값은?


```python
# pandas_data3.txt file read (index_col=0: first column as index)
df1 = pd.read_table('pandas_data3.txt', index_col=0)
df1

# np.sum(): boolean values True=1, False=0 sum
# df1['final']>=40: final score >= 40 condition (True/False)
# Count students with final score >= 40
np.sum(df1['final'] >= 40)

# Filter 1st year students and extract final scores
# df1['year']==1: condition for 1st year students
# loc[condition, 'final']: select rows meeting condition and 'final' column
s1 = df1.loc[df1['year'] == 1, 'final']
s1
list(s1)
np.mean(s1)

# Filter students with proj score == 100 and select specific columns
# df1['proj']==100: condition for proj score 100
# [['mid','final','att']]: select three columns
df1.loc[df1['proj'] == 100, ['mid', 'final', 'att']]

# Multiple conditions: att >= 5 AND proj >= 90
# &: logical AND operator for pandas
# df2: DataFrame meeting both conditions
df2 = df1.loc[(df1['att'] >= 5) & (df1['proj'] >= 90)]
list(df2.index)

# Function to convert mid score to grade
def get_mid_grade(x):
    if x >= 25:
        return 'A'
    elif x >= 20:
        return 'B'
    else:
        return 'C'
    
# Read data again and apply grade function
# .map(function): apply function to each element in Series
d1 = pd.read_table('pandas_data3.txt', index_col=0)
d1.head()

d1['mid_grade'] = d1['mid'].map(get_mid_grade)
d1.head()

# Group by gender and calculate mean of final scores
# .groupby('column'): group data by specified column
# ['final'].mean(): calculate mean of final scores for each group
d1 = pd.read_table('pandas_data3.txt', index_col=0)
d1.groupby('gender')['final'].mean()
```

```python
####### df_2023 전처리하기
df_2023 = pd.read_csv("data/전국평균 분양가격 (2015년10월~2023년6월).csv", encoding="cp949") #default: sep=','
df_2023.head()
df_2023.info()

### ==> 1) '분양가격(제곱미터)'column: nan(결측치) 제거
df_2023_nn=df_2023.loc[df_2023['분양가격(제곱미터)'].notnull()]
#분양가격이 notnull인것만 모아서 다시저장

# ### ==> 2) '분양가격(제곱미터)'column: 숫자로 구성된 문자 -> float형 | 숫자가 아닌 문자 -> nan으로 변경 후 행 제거
df_2023_ft=df_2023_nn.copy()
df_2023_ft['분양가격(제곱미터)']=pd.to_numeric(df_2023_ft['분양가격(제곱미터)'], errors='coerce')
df_2023_ft=df_2023_ft.loc[df_2023_ft['분양가격(제곱미터)'].notnull()]

### ==> 3) '분양가격(제곱미터)'을  '평당분양가격'으로 변환하기
df_2023_ft["평당분양가격"]=df_2023_ft['분양가격(제곱미터)'] * 3.3
df_prep=df_2023_ft.drop(columns=['분양가격(제곱미터)'], errors='ignore')# '분양가격(제곱미터)'' column이 없는 경우 error raise 방지

### ==> 4) '규모구분'을 '전용면적'으로 변환하기
def parse_area_only(x):
    x2=x.replace("모든면적", "전체").replace("전용면적", "").replace("제곱미터", "").replace("60이하", "~60")
    x3=x2.replace("초과", "~").replace("이하", "").replace(" ", "")
    return x3
    
df_prep["전용면적"] = df_prep["규모구분"].map(parse_area_only)
df_prep=df_prep.drop(columns=['규모구분'], errors='ignore') # '규모구분' column이 없는 경우 error raise 방지

## ==> 5) reset_index 후 전처리한 데이터 확인하기
df_prep=df_prep.reset_index(drop=True)

df_prep.head()
df_prep.info()
```

## map함수 활용

* sr1은 int형과 str형 값들이 들어있다. 값이
* 1) int형 또는
  2) 숫자로만 구성된 str형
* 인 경우 1000을 더한 값을, 그렇지 않은 경우 그대로의 값을 갖는 sr2를 출력하시오
* 힌트) map(),  isdigit()

```python
sr1=pd.Series([10, '20', 30, 'ab', 40, '50k'])

## 코드 작성 ##
def func1(x):
    if type(x) == int:
        return x+1000
    elif x.isdigit():
        return int(x)+1000
    else:
        return x

sr2=sr1.map(func1)
sr1
sr2
```
